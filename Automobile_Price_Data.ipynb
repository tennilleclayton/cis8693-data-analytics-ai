```python
# Import pandas; short name "pd"
import pandas as pd

# Import train_test_split.
from sklearn.model_selection import train_test_split

# Import LinearRegression class.
from sklearn.linear_model import LinearRegression
```

```python
data = pd.read_csv("Automobile_Price_Data.csv")
```

```python
# Omitting missing values and modifies the data
data.dropna(inplace=True)

# Checks if there are any missing values;
# The first .any() checks each column.
# The second .any() is then applied to the Series generated by the first .any().
data.isnull().any().any()
```

```python
# Convert current type into a pandas categorical variable and .
columns = ['make', 'fuel-type', 'aspiration', 'num-of-doors', 'body-style',
           'drive-wheels', 'engine-location', 'engine-type',
           'num-of-cylinders', 'fuel-system']

for col in columns:
    data[col] = data[col].astype(pd.CategoricalDtype())
    data[col] = data[col].cat.codes
```

```python
# Checking datatypes
data.dtypes
```

```python
data.describe()
```

```python
# Select Price as the dependent/outcome variable
y = data['price']
```

```python
# Create X, independent/predictor variables
X = data.drop(columns=['price'])
```

```python
# Split X and y into training data (80%) and testing data (20%).
train_X, test_X, train_y, test_y = train_test_split(X, y, test_size=0.2,
                                                    random_state=1)
```

```python
# Create a LinearRegression model object.
model = LinearRegression()
```

```python
# Fit/train the model using the predictor values in train_X
# and the known Price values in train_y.
model.fit(train_X, train_y)
```

```python
# Print the intercept (constant term) of the fitted linear regression equation.
print('intercept ', model.intercept_)
```

```python
# Create and print a table that matches each predictor name with its estimated regression coefficient
print(pd.DataFrame({'Predictor': X.columns, 'coefficient': model.coef_}))
```

```python
# Calculate R-squared on the training data; it measures how much variation in Price the model explains in the training set.
r2_train = model.score(train_X, train_y)

# Display the training R-squared value in a notebook/interactive environment.
r2_train
```

```python
# Calculate R-squared on the testing data to evaluate how well the model performs on data it did not train on.
r2_test = model.score(test_X, test_y)

# Display the testing R-squared value in a notebook/interactive environment.
r2_test
```
