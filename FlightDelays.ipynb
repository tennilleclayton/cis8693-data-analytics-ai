```python
import dmba
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
from dmba import classificationSummary
```

```python
flightdelays_df = pd.read_csv('FlightDelays.csv') #read data
flightdelays_df.head()
```

```python
flightdelays_df.columns = [c.replace(' ', '_') for c in flightdelays_df.columns]
# Displays the first five rows of the dataset so we can inspect the data.
flightdelays_df.head()
```

```python
# Returns number of observations, number of varialbes
flightdelays_df.shape
```

```python
# Count missing values
flightdelays_df.isnull().sum()
```

```python
# View Data Types
flightdelays_df.dtypes
```

```python
# Convert current type into a pandas categorical variable and .
columns = ['CARRIER', 'DEST', 'ORIGIN', 'Flight_Status']

for col in columns:
    flightdelays_df[col] = flightdelays_df[col].astype(pd.CategoricalDtype())
    flightdelays_df[col] = flightdelays_df[col].cat.codes
```

```python
flightdelays_df.head()
```

```python
# Selects Flight_Status as the outcome variable that the logistic regression model will predict.
y = flightdelays_df['Flight_Status'] #define the outcome variable
# Creates the predictor dataset X by keeping all columns except the outcome variable Flight_Status.
X = flightdelays_df.drop(columns=['Flight_Status']) #define the predictors

# partition data
# Splits X and y into 70% training data and 30% testing data; random_state=1 makes the split reproducible.
train_X, test_X, train_y, test_y = train_test_split(X, y, test_size=0.3, random_state=1)
```

```python
# Creates a logistic regression model and stores it in the variable logit_reg.
logit_reg = LogisticRegression()
# Trains the logistic regression model using the training predictors and training outcome values.
logit_reg.fit(train_X, train_y)
```

```python
# Prints the intercept estimated by the logistic regression model.
print('intercept ', logit_reg.intercept_[0])
# Creates and prints a table showing the estimated coefficient for each predictor variable.
print(pd.DataFrame({'coeff': logit_reg.coef_[0]}, index=X.columns).transpose())
```

```python
# Evaluates the model on the training data by comparing actual outcomes with predicted outcomes. (predicted value)
classificationSummary(train_y, logit_reg.predict(train_X))
```

```python
# Evaluates the model on the testing data by comparing actual outcomes with predicted outcomes. (predicted value)
classificationSummary(test_y, logit_reg.predict(test_X))
```
