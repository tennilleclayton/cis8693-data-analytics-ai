```python
# Imports pandas as pd so we can load and work with tabular data.
import pandas as pd
# Imports the decision-tree classifier used when the dependent variable is categorical or binary.
from sklearn.tree import DecisionTreeClassifier #if your dependent variable is binary
# Imports the decision-tree regressor used when the dependent variable is numerical.
from sklearn.tree import DecisionTreeRegressor  #if your dependent variable is numerical, like Sales
# Imports the random-forest classifier, which combines many decision trees for classification.
from sklearn.ensemble import RandomForestClassifier
# Imports the function used to split data into training and testing sets.
from sklearn.model_selection import train_test_split
# Imports the dmba package.
import dmba
# Imports dmba helper functions for plotting decision trees and summarizing classification and regression results.
from dmba import plotDecisionTree, classificationSummary, regressionSummary
```

```python
# Reads the GermanCredit.csv file into a pandas DataFrame named GermanCredit_df.
GermanCredit_df = pd.read_csv('GermanCredit.csv')
# Displays the first five rows of GermanCredit_df so we can preview the data.
GermanCredit_df.head()
```

```python
# Creates a decision-tree classifier and fixes the random seed for reproducible results.
classTree = DecisionTreeClassifier(random_state=0)
# Trains the decision tree using all columns except RESPONSE as predictors and RESPONSE as the target.
classTree.fit(GermanCredit_df.drop(columns=['RESPONSE']), GermanCredit_df['RESPONSE'])
```

```python
# Creates the predictor matrix X by removing the RESPONSE target column from the dataset.
X = GermanCredit_df.drop(columns=['RESPONSE'])
# Creates the target vector y using the RESPONSE column.
y = GermanCredit_df['RESPONSE']
```

```python
# Splits X and y into 80% training data and 20% testing data, using a fixed random seed.
train_X, test_X, train_y, test_y = train_test_split(X, y, test_size=0.2, random_state=1)
```

```python
# Creates an unrestricted decision-tree classifier with default settings.
fullClassTree = DecisionTreeClassifier(random_state=0)
# Trains the full decision tree on the training predictors and training target.
fullClassTree.fit(train_X, train_y)
```

```python
# Evaluates the full decision tree on the training data by comparing actual and predicted classes.
classificationSummary(train_y, fullClassTree.predict(train_X))
```

```python
# Evaluates the full decision tree on the test data by comparing actual and predicted classes.
classificationSummary(test_y, fullClassTree.predict(test_X))
```

```python
# Creates a smaller decision tree with maximum depth 10.
smallClassTree = DecisionTreeClassifier(max_depth=10)
# Trains the smaller decision tree on the training data.
smallClassTree.fit(train_X, train_y)
```

```python
# Evaluates the smaller decision tree on the training data.
classificationSummary(train_y, smallClassTree.predict(train_X))
```

```python
# Evaluates the smaller decision tree on the test data.
classificationSummary(test_y, smallClassTree.predict(test_X))
```

```python
# Creates a random-forest classifier containing 500 decision trees and fixes the random seed.
# random_state should always be 1 to give you the same results
rf = RandomForestClassifier(n_estimators=500, random_state=1)
# Trains the random-forest classifier on the training data.
rf.fit(train_X, train_y)
```

```python
# Evaluates the random forest on the training data by comparing actual and predicted classes.
classificationSummary(train_y, rf.predict(train_X))
```

```python
# Evaluates the random forest on the test data by comparing actual and predicted classes.
classificationSummary(test_y, rf.predict(test_X))
```
